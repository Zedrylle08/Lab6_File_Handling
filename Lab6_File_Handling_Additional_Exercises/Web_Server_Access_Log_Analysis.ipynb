```python
from pathlib import Path

WORK_DIR = Path.cwd()
if WORK_DIR.name != "Lab6_File_Handling_Additional_Exercises":
    candidate = WORK_DIR / "Lab6_File_Handling_Additional_Exercises"
    if candidate.is_dir():
        WORK_DIR = candidate

LAST_NAME = "Adonis"
STUDENT_ID = "TUPM-26-1828"
SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(digit) for digit in STUDENT_ID if digit.isdigit())

DATA_DIR = WORK_DIR / "data"
OUTPUT_DIR = WORK_DIR / "output" / "web_log_analysis"
BACKUP_DIR = WORK_DIR / "backup" / "web_log_analysis"
for folder in (DATA_DIR, OUTPUT_DIR, BACKUP_DIR):
    folder.mkdir(parents=True, exist_ok=True)

print("Student:", LAST_NAME)
print("Student ID:", STUDENT_ID)
print("Seed digit:", SEED_DIGIT)
print("ID sum:", ID_SUM)
print("Exercise folder:", WORK_DIR.resolve())
```

```python
import csv
import json
import shutil
import urllib.request
import zipfile
from collections import Counter

dataset_url = "https://www.kaggle.com/api/v1/datasets/download/shawon10/web-log-dataset"
archive_path = DATA_DIR / "web-log-dataset.zip"
dataset_path = DATA_DIR / "weblog.csv"

if not archive_path.exists():
    urllib.request.urlretrieve(dataset_url, archive_path)
if not dataset_path.exists():
    with zipfile.ZipFile(archive_path) as archive:
        with archive.open("weblog.csv") as source:
            with open(dataset_path, "wb") as destination:
                shutil.copyfileobj(source, destination)

sample_size = 1000 + ID_SUM
target_status_class = f"{(SEED_DIGIT % 4) + 2}xx"
filtered_csv_path = OUTPUT_DIR / f"{LAST_NAME}_{STUDENT_ID}_{target_status_class}_requests.csv"
summary_json_path = OUTPUT_DIR / f"{LAST_NAME}_{STUDENT_ID}_web_log_summary.json"
backup_json_path = BACKUP_DIR / summary_json_path.name

status_counts = Counter()
target_url_counts = Counter()
records_processed = 0
valid_records = 0
invalid_records = 0
records_in_target_class = 0

with open(dataset_path, "r", encoding="utf-8", errors="replace", newline="") as source:
    reader = csv.reader(source)
    header = next(reader, None)
    if not header:
        raise ValueError("The web log CSV is empty or has no header.")
    columns = {name.strip().lower(): index for index, name in enumerate(header)}
    status_column = columns.get("status", columns.get("staus"))
    required_columns = {"ip", "time", "url"}
    if not required_columns.issubset(columns) or status_column is None:
        raise ValueError(f"Unexpected web log columns: {header}")

    output_fields = ["IP", "Time", "URL", "Status"]
    with open(filtered_csv_path, "w", encoding="utf-8", newline="") as filtered_file:
        writer = csv.DictWriter(filtered_file, fieldnames=output_fields)
        writer.writeheader()
        for _ in range(sample_size):
            row = next(reader, None)
            if row is None:
                break
            records_processed += 1
            try:
                ip = row[columns["ip"]].strip()
                timestamp = row[columns["time"]].strip()
                request = row[columns["url"]].strip()
                status = int(row[status_column].strip())
                if not ip or not timestamp or not request or not 100 <= status <= 599:
                    raise ValueError("Missing or invalid log field")
                request_parts = request.split()
                requested_url = request_parts[1] if len(request_parts) >= 2 else request
                if not requested_url:
                    raise ValueError("Missing requested URL")
            except (IndexError, ValueError):
                invalid_records += 1
                continue

            valid_records += 1
            status_counts[str(status)] += 1
            if status // 100 == int(target_status_class[0]):
                records_in_target_class += 1
                target_url_counts[requested_url] += 1
                writer.writerow({"IP": ip, "Time": timestamp, "URL": requested_url, "Status": status})

summary = {
    "last_name": LAST_NAME,
    "student_id": STUDENT_ID,
    "records_requested": sample_size,
    "records_processed": records_processed,
    "valid_records": valid_records,
    "invalid_records": invalid_records,
    "target_status_class": target_status_class,
    "records_in_target_class": records_in_target_class,
    "status_code_counts": dict(sorted(status_counts.items(), key=lambda item: int(item[0]))),
    "top_5_urls_in_target_class": [
        {"url": url, "requests": count}
        for url, count in target_url_counts.most_common(5)
    ],
    "filtered_csv_file": str(filtered_csv_path),
    "summary_json_file": str(summary_json_path),
    "backup_json_file": str(backup_json_path),
    "backup_verified": False,
}
with open(summary_json_path, "w", encoding="utf-8") as summary_file:
    json.dump(summary, summary_file, indent=4)
shutil.copy2(summary_json_path, backup_json_path)
summary["backup_verified"] = summary_json_path.read_bytes() == backup_json_path.read_bytes()
if not summary["backup_verified"]:
    raise OSError("JSON summary backup could not be verified.")
with open(summary_json_path, "w", encoding="utf-8") as summary_file:
    json.dump(summary, summary_file, indent=4)
shutil.copy2(summary_json_path, backup_json_path)
if summary_json_path.read_bytes() != backup_json_path.read_bytes():
    raise OSError("JSON summary backup does not match the final summary.")

print("Records processed:", records_processed)
print("Valid records:", valid_records)
print("Invalid records:", invalid_records)
print("Target status class:", target_status_class)
print("Records in target class:", records_in_target_class)
print("Count per status code:", dict(sorted(status_counts.items(), key=lambda item: int(item[0]))))
print("Five most requested URLs:", summary["top_5_urls_in_target_class"])
print("Filtered CSV:", filtered_csv_path)
print("JSON summary:", summary_json_path)
print("Backup verified:", summary["backup_verified"])
```
