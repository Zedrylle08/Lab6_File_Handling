```python
from pathlib import Path
import shutil
import zipfile

WORK_DIR = Path.cwd()
if WORK_DIR.name != "Lab6_File_Handling_Additional_Exercises":
    candidate = WORK_DIR / "Lab6_File_Handling_Additional_Exercises"
    if candidate.is_dir():
        WORK_DIR = candidate

LAST_NAME = "Adonis"
STUDENT_ID = "TUPM-26-1828"
SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(digit) for digit in STUDENT_ID if digit.isdigit())
OPERATING_SYSTEM = "Android" if SEED_DIGIT % 2 == 0 else "iOS"
TARGET_BEHAVIOR_CLASS = (SEED_DIGIT % 5) + 1

DATA_DIR = WORK_DIR / "data"
OUTPUT_DIR = WORK_DIR / "output" / "mobile_device_usage"
BACKUP_DIR = WORK_DIR / "backup" / "mobile_device_usage"
for folder in (DATA_DIR, OUTPUT_DIR, BACKUP_DIR):
    folder.mkdir(parents=True, exist_ok=True)

dataset_url = "https://www.kaggle.com/api/v1/datasets/download/valakhorasani/mobile-device-usage-and-user-behavior-dataset"
archive_path = DATA_DIR / "mobile-device-usage-dataset.zip"
dataset_path = DATA_DIR / "user_behavior_dataset.csv"
if not archive_path.exists():
    from urllib.request import urlretrieve
    urlretrieve(dataset_url, archive_path)
if not dataset_path.exists():
    with zipfile.ZipFile(archive_path) as archive:
        with archive.open("user_behavior_dataset.csv") as source:
            with open(dataset_path, "wb") as destination:
                shutil.copyfileobj(source, destination)

print("Student:", LAST_NAME)
print("Student ID:", STUDENT_ID)
print("Operating system:", OPERATING_SYSTEM)
print("Target behavior class:", TARGET_BEHAVIOR_CLASS)
print("Dataset:", dataset_path)
```

```python
import csv
import json

export_fields = [
    "User ID",
    "Device Model",
    "Operating System",
    "Battery Drain (mAh/day)",
    "Data Usage (MB/day)",
]
filtered_csv_path = OUTPUT_DIR / f"{LAST_NAME}_{STUDENT_ID}_mobile_users.csv"
summary_json_path = OUTPUT_DIR / f"{LAST_NAME}_{STUDENT_ID}_mobile_summary.json"
backup_csv_path = BACKUP_DIR / filtered_csv_path.name

total_records = 0
matching_os_records = 0
selected_records = 0
invalid_records = 0
battery_total = 0.0
data_usage_total = 0.0

with open(dataset_path, "r", encoding="utf-8-sig", newline="") as source:
    reader = csv.DictReader(source)
    required_fields = set(export_fields + ["User Behavior Class"])
    if not reader.fieldnames or not required_fields.issubset(reader.fieldnames):
        raise ValueError(f"Unexpected dataset columns: {reader.fieldnames}")
    with open(filtered_csv_path, "w", encoding="utf-8", newline="") as output:
        writer = csv.DictWriter(output, fieldnames=export_fields)
        writer.writeheader()
        for row in reader:
            total_records += 1
            if row["Operating System"].strip() != OPERATING_SYSTEM:
                continue
            matching_os_records += 1
            try:
                behavior_class = int(row["User Behavior Class"].strip())
            except (AttributeError, ValueError):
                invalid_records += 1
                continue
            if behavior_class != TARGET_BEHAVIOR_CLASS:
                continue
            try:
                battery_drain = float(row["Battery Drain (mAh/day)"].strip())
                data_usage = float(row["Data Usage (MB/day)"].strip())
                if not row["User ID"].strip() or not row["Device Model"].strip():
                    raise ValueError("Missing user or device field")
            except (AttributeError, ValueError):
                invalid_records += 1
                continue
            writer.writerow({field: row[field] for field in export_fields})
            selected_records += 1
            battery_total += battery_drain
            data_usage_total += data_usage

average_battery_drain = battery_total / selected_records if selected_records else 0.0
average_data_usage = data_usage_total / selected_records if selected_records else 0.0
with open(filtered_csv_path, "r", encoding="utf-8", newline="") as exported_file:
    exported_row_count = sum(1 for _ in csv.DictReader(exported_file))
count_verified = exported_row_count == selected_records
if not count_verified:
    raise ValueError("Exported CSV row count does not match the selected record count.")

summary = {
    "last_name": LAST_NAME,
    "student_id": STUDENT_ID,
    "operating_system": OPERATING_SYSTEM,
    "target_behavior_class": TARGET_BEHAVIOR_CLASS,
    "total_records_in_dataset": total_records,
    "records_matching_operating_system": matching_os_records,
    "records_in_target_behavior_class": selected_records,
    "invalid_records_skipped": invalid_records,
    "average_battery_drain_mah_per_day": average_battery_drain,
    "average_data_usage_mb_per_day": average_data_usage,
    "exported_csv_row_count": exported_row_count,
    "count_verified": count_verified,
    "exported_csv_file": str(filtered_csv_path),
    "summary_json_file": str(summary_json_path),
    "backup_csv_file": str(backup_csv_path),
}
with open(summary_json_path, "w", encoding="utf-8") as summary_file:
    json.dump(summary, summary_file, indent=4)
shutil.copy2(filtered_csv_path, backup_csv_path)
backup_size_bytes = backup_csv_path.stat().st_size
backup_verified = (
    backup_csv_path.exists()
    and backup_size_bytes == filtered_csv_path.stat().st_size
    and backup_csv_path.read_bytes() == filtered_csv_path.read_bytes()
)
if not backup_verified:
    raise OSError("Exported CSV backup could not be verified.")
summary["backup_size_bytes"] = backup_size_bytes
summary["backup_verified"] = backup_verified
with open(summary_json_path, "w", encoding="utf-8") as summary_file:
    json.dump(summary, summary_file, indent=4)

print("Operating system and behavior class:", OPERATING_SYSTEM, TARGET_BEHAVIOR_CLASS)
print("Total records in dataset:", total_records)
print("Records matching operating system:", matching_os_records)
print("Records in target behavior class:", selected_records)
print("Average battery drain (mAh/day):", round(average_battery_drain, 2))
print("Average data usage (MB/day):", round(average_data_usage, 2))
print("Invalid records skipped:", invalid_records)
print("Count verification:", count_verified, f"({exported_row_count} exported rows)")
print("Exported CSV:", filtered_csv_path)
print("JSON summary:", summary_json_path)
print("Backup:", backup_csv_path)
print("Backup size (bytes):", backup_size_bytes)
print("Backup verified:", backup_verified)
```
