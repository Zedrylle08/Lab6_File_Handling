```python
from pathlib import Path

LAST_NAME = "Adonis"
STUDENT_ID = "TUPM-26-1828"

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGTH = len(LAST_NAME)

print("Seed:", SEED_DIGIT)
print("ID Sum:", ID_SUM)
print("Name Length:", NAME_LENGTH)
```

```python
WORK_DIR = Path(".")
DATA_DIR = WORK_DIR / "data"
OUTPUT_DIR = WORK_DIR / "output"
BACKUP_DIR = WORK_DIR / "backup"

DATA_DIR.mkdir(exist_ok=True)
OUTPUT_DIR.mkdir(exist_ok=True)
BACKUP_DIR.mkdir(exist_ok=True)

print("Student:", LAST_NAME)
print("Student ID:", STUDENT_ID)
print("Workspace:", WORK_DIR.resolve())
print("Data folder exists:", DATA_DIR.exists())
print("Output folder exists:", OUTPUT_DIR.exists())
print("Backup folder exists:", BACKUP_DIR.exists())
```

```python
file_path = WORK_DIR / f"{LAST_NAME}_{STUDENT_ID}_notes.txt"print("Full path:", file_path)print("File name:", file_path.name)print("File extension:", file_path.suffix)print("Parent folder:", file_path.parent)with open(file_path, "w", encoding="utf-8") as file:    file.write(f"Student: {LAST_NAME}\n")    file.write(f"Student ID: {STUDENT_ID}\n")    file.write(f"Seed digit: {SEED_DIGIT}\n")    file.write(f"ID sum: {ID_SUM}\n")print("Created:", file_path)print("Exists:", file_path.exists())
```

```python
verification_value = ID_SUM * NAME_LENGTH + SEED_DIGITwith open(file_path, "a", encoding="utf-8") as file:    file.write(f"Verification value: {verification_value}\n")with open(file_path, "r", encoding="utf-8") as file:    print(file.read())
```

```python
filtered_path = OUTPUT_DIR / f"{LAST_NAME}_{STUDENT_ID}_filtered.txt"with open(file_path, "r", encoding="utf-8") as source:    with open(filtered_path, "w", encoding="utf-8") as output:        for line in source:            if "ID" in line or "Verification" in line:                output.write(line)print("Filtered file:", filtered_path)print("Contents:")with open(filtered_path, "r", encoding="utf-8") as file:    print(file.read())
```

```python
csv_path = OUTPUT_DIR / f"{LAST_NAME}_{STUDENT_ID}_record.csv"with open(csv_path, "w", encoding="utf-8") as file:    file.write("last_name,student_id,seed_digit,id_sum\n")    file.write(f"{LAST_NAME},{STUDENT_ID},{SEED_DIGIT},{ID_SUM}\n")print("Created:", csv_path)with open(csv_path, "r", encoding="utf-8") as file:    header = file.readline().strip().split(",")    record = file.readline().strip().split(",")print("Fields:", header)print("Values:", record)for field, value in zip(header, record):    print(f"{field}: {value}")
```

```python
import jsonjson_path = OUTPUT_DIR / f"{LAST_NAME}_{STUDENT_ID}_record.json"student_data = {    "last_name": LAST_NAME,    "student_id": STUDENT_ID,    "seed_digit": SEED_DIGIT,    "id_sum": ID_SUM,    "name_length": NAME_LENGTH,    "verification_value": verification_value,}with open(json_path, "w", encoding="utf-8") as file:    json.dump(student_data, file, indent=4)print("Created:", json_path)with open(json_path, "r", encoding="utf-8") as file:    data = json.load(file)expected_value = data["id_sum"] * data["name_length"] + data["seed_digit"]print("Stored verification:", data["verification_value"])print("Expected verification:", expected_value)print("Status:", "VERIFIED" if data["verification_value"] == expected_value else "CHECK")
```

```python
import shutilfile_info = csv_path.stat()print("File name:", csv_path.name)print("File size:", file_info.st_size, "bytes")print("File extension:", csv_path.suffix)print("File exists:", csv_path.exists())backup_path = BACKUP_DIR / csv_path.nameshutil.copy2(csv_path, backup_path)print("Original:", csv_path)print("Backup:", backup_path)print("Backup exists:", backup_path.exists())
```

```python
import shutilimport urllib.requestimport zipfiledataset_url = "https://archive.ics.uci.edu/static/public/235/individual+household+electric+power+consumption.zip"archive_path = DATA_DIR / "individual_household_electric_power_consumption.zip"dataset_path = DATA_DIR / "household_power_consumption.txt"if not archive_path.exists():    urllib.request.urlretrieve(dataset_url, archive_path)if not dataset_path.exists():    with zipfile.ZipFile(archive_path) as archive:        with archive.open("household_power_consumption.txt") as source:            with open(dataset_path, "wb") as destination:                shutil.copyfileobj(source, destination)print("Dataset:", dataset_path)print("File exists:", dataset_path.exists())with open(dataset_path, "r", encoding="utf-8") as file:    header = file.readline().strip()    records = [file.readline().strip() for _ in range(3)]print("Header:")print(header)print("\\nSample records:")for record in records:    print(record)
```

```python
sample_size = 50 + ID_SUMtotal_power = 0.0valid_records = 0with open(dataset_path, "r", encoding="utf-8") as file:    next(file)    for _ in range(sample_size):        line = file.readline()        if not line:            break        fields = line.strip().split(";")        if len(fields) > 2 and fields[2] != "?":            total_power += float(fields[2])            valid_records += 1average_power = total_power / valid_records if valid_records else 0print("Sample size:", sample_size)print("Valid records:", valid_records)print("Average Global Active Power:", round(average_power, 3))
```
